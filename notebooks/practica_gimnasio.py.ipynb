# Modulo 3. Diagnostico y Manejo de datos faltantes.

Se identifican y corrigen los valores nulos utilizando media, mediana, moda y valores constantes.

```python
print("Diagnostico de datos faltantes")
print(df.isnull().sum())
print("\nPorcentaje de valores faltantes")
print((df.isnull().sum() / len(df) *100).round(2))
```

## Trabajando con los valores faltantes


```python
df['Edad'] = df['Edad'].fillna(df['Edad'].median())
df['Sucursal'] = df['Sucursal'].fillna(df['Sucursal'].mode()[0])
df = df.drop(columns='Email')
print("\nNulos Restantes:")
print(df.isnull().sum(), df.shape)
```

# Modulo 4. Detección de Duplicados

```python
#Copiar el dataset a una variable para trabajar con los datos duplicados.

datos_duplicados = df[df.duplicated(keep=False)]

#Ordenar sucursal para ver el nombre, edad y plan.
print("\nFilas antes: ", len(df))
print("\n---Historial completo de duplicados encontrados ordenados por sucursal---")
print(f'\n{datos_duplicados.sort_values(by='Sucursal')}')
```

```python
#Procedemos a eliminar los datos duplicados.
df.drop_duplicates(inplace=True)
print(df.duplicated())
print("Filas despues de eliminar duplicados: ", len(df))
```

# Modulo 5. Normalización y manejo de tipo de datos.

```python
#Trabajando con tipos de datos
df['Edad'] = df['Edad'].astype(int)
df['Plan'] = df['Plan'].astype('category')
print(df.dtypes)
```

# Tres preguntas para cerrar

### 1. Mediana vs. promedio, con datos reales
**Calcula la media y la mediana de Edad incluyendo a los clientes mayores de 85 años, y luego calcúlalas otra vez
excluyéndolos. ¿Cuál de las dos medidas cambió más al quitar esos casos? ¿Por qué tiene sentido ese resultado?**

La media cambio mas que la mediana al eliminar los clientes mayores de 85 años. Esto ocurre porque la media utiliza todos los valores del dataframe y es sensible a los valores extremos. En cambio, la mediana representa el valor central de los datos ordenados y apenas se ve afectada por la eliminacion de unos pocos valores altos.

### 2. Dos columnas, dos estrategias distintas
**Sucursal tiene apenas 3.17% de nulos y Email tiene 74.60%. ¿Por qué la estrategia correcta no es la misma para
ambas, aunque las dos sean columnas de texto?**

En la columan 'Sucursal' solo faltan unos pocos datos, asi que podemos completarlo facilmente y seguir aprovechando esa informacion. Sin embargo, en la columna 'Email' faltan la mayoria de los registros. Si intentaramos rellenarlos, estariamos adivinando demasiados datos y eso haria que el analisis fuera poco confiable. Por esta razon, lo mejor es eiliminar esa columna y trabajar con informacion de mejor calidad.

### 3. Mirando hacia adelante
***Si este gimnasio quisiera más adelante predecir qué clientes están en riesgo de cancelar su membresía, ¿qué
precaución tomarías con los datos que acabas de limpiar antes de usarlos para eso?***

Aunque los datos ya estan limpios, antes de usarlos para hacer predicciones verificaria que contengan la informacion necesaria para entrenar el modelo. En este caso seria indispensable contar con mas atributos o features que indiquen, por ejemplo, si el cliente cancelo o no la membresia ya que esa sera la varaiable que el modelo aprendera a predecir. Tambien revisaria que los datos sigan siendo representativos de la realidad y que variables importantes como los dias de asistencia no se hayan perdido durante el preoceso de limpieza.


```python
mediana_original = df['Edad'].median()
media_original = df['Edad'].mean()

df_sin85 = df[df['Edad'] <= 85]

media_sin85 = df_sin85['Edad'].mean()
mediana_sin85 = df_sin85['Edad'].median()

print("----Resultados Con todas las edades------")
print("Media: ", media_original)
print("Mediana: ", mediana_original)

print("\n----Resultados sin edades mayores de 85----")
print("Media: ", media_sin85)
print("Mediana: ", mediana_sin85)
```
