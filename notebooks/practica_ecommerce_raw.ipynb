# Modulo 3. Diagnostico y Manejo de datos faltantes.

Se identifican y corrigen los valores nulos utilizando media, mediana, moda y valores constantes.

```python
print("Diagnostico de datos faltantes: ")
print(df.isnull().sum())
print("\nPorcentaje de datos faltantes")
print((df.isnull().sum() / len(df) * 100).round(2) )
```

## Trabajando con los valores faltantes


```python
#Imputar con mediana la columna de TiempoEntregaDias
tiempo_entrega_faltantes = df['TiempoEntregaDias'].median()
df['TiempoEntregaDias'] = df['TiempoEntregaDias'].fillna(tiempo_entrega_faltantes)

#Imputar con moda la columna 'Metodo Pago'
metodo_pago_faltantes = df['MetodoPago'].mode()[0]
df['MetodoPago'] = df['MetodoPago'].fillna(metodo_pago_faltantes)

#Eliminar la columna de 'ComentarioCliente'
df = df.drop(columns='ComentarioCliente')

#Imprimir resultados
print("Nulos, restantes", df.isnull().sum().sum())

```

# Modulo 4. Detección de Duplicados

```python
copia_dataframe = df
print("Cantidad de filas antes de eliminar duplicados: ", len(copia_dataframe))
df = df.drop_duplicates()
print("Cantidad de filas despues de eliminar duplicados:", len(df))
```

# Modulo 5. Normalización y manejo de tipo de datos.

```python
#Trabando con tipos de datos
df['CategoriaProducto'] = df['CategoriaProducto'].astype('category')
print(df.dtypes)
print(df.isnull().sum(), df.shape)
```

# Modulo 6. Ingeniería de Características

```python
#combinar columnas - Monto total
df['MontoTotal'] = (df['CantidadArticulos'] * df['PrecioUnitario']).round(2)
print(df[['CantidadArticulos', 'PrecioUnitario', 'MontoTotal']].head())


```

```python
#Bandera Binaria - PedidoGrande.
print(df['MontoTotal'].describe())

umbral = df['MontoTotal'].median().round(2)
df['PedidosGrande'] = (df['MontoTotal'] > umbral).astype(int)

print(f'\n Pedidos Menores y Grandes  {df['PedidosGrande'].value_counts()}')
print("\n Visualizacion de Pedidos Grandes")
print(df[['CategoriaProducto', 'MontoTotal', 'PedidosGrande']].head())
```

```python
#Extraer texto con regex - Sector

df['Sector'] = df['DireccionEnvio'].str.extract(r',\s*(.+)$')

print(df['Sector'].value_counts())
print()
print(df[['DireccionEnvio', 'Sector']].head(15))
```

```python
#Agrupar en Categorias VelocidadEntrega

df['VelocidadEntrega'] = pd.cut(
    df['TiempoEntregaDias'],
    bins=[0, 5, 15, 30],
    labels=('Rapida', 'Normal', 'Lenta')
)

print(df[['TiempoEntregaDias', 'VelocidadEntrega']].head(30))
print()
print(df['VelocidadEntrega'].value_counts())
```

```python
print("="*50)
print("RESUMEN FINAL DEL DATASET")
print("="*50)
print(f'Filas: {df.shape[0]}')
print(f'Columnas: {df.shape[1]}')
print(f'\nColumnas nuevas creadas hoy:')
print([c for c in df.columns if c in ['MontoTotal', 'PedidosGrande', 'Sector', 'VelocidadEntrega']])
print(f'\nValores nulos restantes: {df.isnull().sum().sum()}')

```
